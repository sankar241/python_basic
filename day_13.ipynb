```python
#mathemetical operations on sets
#these operations

#union(|):combine all unique elements from both sets
a={1,2,3,4,5}
b={3,5,7,8,9,0}
print(a|b)
```

```python

#intersection(&)
#returns only the common elements from both sets
a={1,2,3,4,5}
b={3,5,7,8,9,0}
print(a & b)
```

```python
#difference(-):a lo common values kakunda remaining values
print(a-b)
```

```python
#symmetric difference(^)
#returns elements thet are present in either set but not in both
a={1,2,3,4,5}
b={3,5,7,8,9,0}
print(a^b)
```
